```python
import pandas as pd
import numpy as np

df = pd.read_csv(
    "../data/raw data/global air pollution dataset.csv"
)

df.head()
```

```python
clean_df = df.copy()
```

```python
clean_df
```

```python
clean_df.columns = (
    clean_df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace(".", "_", regex=False)
)

clean_df.columns
```

```python
text_columns = clean_df.select_dtypes(include="object").columns

for col in text_columns:
    clean_df[col] = clean_df[col].str.strip()
```

```python
clean_df.duplicated().sum()
```

```python
clean_df.isnull().sum()
```

```python
missing_country = clean_df[clean_df["country"].isna()].copy()

missing_country[["city", "country"]]
```
