```python
import pandas as pd
import joblib
```

```python
model = joblib.load(
    "../models/aqi_category_model.pkl"
)

print("Model loaded successfully!")
```

```python
new_data = pd.DataFrame({
    "country": ["India"],
    "city": ["Mumbai"],
    "co_aqi_value": [20],
    "ozone_aqi_value": [40],
    "no2_aqi_value": [30],
    "pm2_5_aqi_value": [150]
})

new_data
```

```python
prediction = model.predict(new_data)

print("Predicted AQI Category:", prediction[0])
```

```python
probabilities = model.predict_proba(new_data)

probability_df = pd.DataFrame(
    probabilities,
    columns=model.classes_
)

probability_df
```
